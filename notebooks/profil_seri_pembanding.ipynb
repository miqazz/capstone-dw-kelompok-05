```python
from pathlib import Path

folder_raw = Path("../data/raw")

for folder in sorted(folder_raw.iterdir()):
    if folder.is_dir():
        print(f"\nFolder: {folder.name}")
        for file in sorted(folder.rglob("*")):
            if file.is_file():
                print(" -", file.relative_to(folder_raw))
```

```python
import pandas as pd
from pathlib import Path

folder_who = Path("../data/raw/who_gho/2026-10-20/datadot")

file_who = {
    "UHC_INDEX_REPORTED": "9A706FD_UHC_INDEX_REPORTED.csv",
    "FINPROTECTION_CATA_TOT_10_POP": "A65146D_FINPROTECTION_CATA_TOT_10_POP.csv",
    "HWF_0001": "217795A_HWF_0001.csv",
    "HWF_0006": "5C8435F_HWF_0006.csv"
}

data_who = {}

for nama, file in file_who.items():
    df = pd.read_csv(folder_who / file)
    data_who[nama] = df

    print("=" * 60)
    print("SERI:", nama)
    print("Ukuran data:", df.shape)
    print("\nNama kolom:")
    print(df.columns.tolist())

    print("\n5 baris pertama:")
    display(df.head())
```

```python
for nama, df in data_who.items():
    print("\n" + "=" * 60)
    print("SERI:", nama)
    print("Jumlah baris:", len(df))

    for kolom in df.columns:
        print(f"\n{kolom}")
        print("Jumlah nilai unik:", df[kolom].nunique())

        if df[kolom].nunique() <= 20:
            print("Nilai:", df[kolom].dropna().unique())
```

```python
for nama, df in data_who.items():
    print("\n" + "=" * 60)
    print("SERI:", nama)

    tahun = sorted(df["DIM_TIME"].dropna().unique())

    print("Periode tersedia:", tahun)
    print("Rentang tahun:", min(tahun), "-", max(tahun))

    kolom_dimensi = [
        kolom for kolom in df.columns
        if kolom.startswith("DIM_") and kolom != "DIM_TIME"
    ]

    print("\nRincian/dimensi:")
    for kolom in kolom_dimensi:
        print(kolom, ":", df[kolom].dropna().unique().tolist())

    kolom_nilai = [
        kolom for kolom in df.columns
        if kolom not in [
            "IND_ID", "IND_CODE", "IND_UUID", "IND_PER_CODE",
            "DIM_TIME", "DIM_TIME_TYPE", "DIM_GEO_CODE_M49",
            "DIM_GEO_CODE_TYPE", "DIM_PUBLISH_STATE_CODE",
            "IND_NAME", "GEO_NAME_SHORT"
        ] + kolom_dimensi
    ]

    print("\nKolom nilai:", kolom_nilai)
```

```python
import json

file_metadata = Path(
    "../data/raw/who_gho/2026-10-20/datadot/ddiindicators_skip0000.json"
)

with open(file_metadata, "r", encoding="utf-8") as f:
    metadata = json.load(f)

print("Tipe metadata:", type(metadata))

if isinstance(metadata, dict):
    print("Kunci metadata:", list(metadata.keys()))
elif isinstance(metadata, list):
    print("Jumlah item:", len(metadata))
    if len(metadata) > 0:
        print("Contoh item pertama:")
        print(metadata[0])
```

```python
# Membaca kembali dataset utama BPS
folder_bps = Path("../data/raw/bps/2026-10-20")

bps_1402 = pd.read_csv(folder_bps / "sdg3_3_8_1_a_v1402.csv")
bps_2279 = pd.read_csv(folder_bps / "sdg3_3_8_2_a_v2279.csv")
bps_1477 = pd.read_csv(folder_bps / "sdg3_3_c_1_v1477.csv")

# Menghitung irisan tahun aktual
perbandingan_tahun = {
    "UHC_INDEX_REPORTED vs var1402": (
        data_who["UHC_INDEX_REPORTED"]["DIM_TIME"],
        bps_1402["tahun"]
    ),
    "FINPROTECTION vs var2279": (
        data_who["FINPROTECTION_CATA_TOT_10_POP"]["DIM_TIME"],
        bps_2279["tahun"]
    ),
    "HWF_0001 vs var1477": (
        data_who["HWF_0001"]["DIM_TIME"],
        bps_1477["tahun"]
    ),
    "HWF_0006 vs var1477": (
        data_who["HWF_0006"]["DIM_TIME"],
        bps_1477["tahun"]
    )
}

for nama, (tahun_who, tahun_bps) in perbandingan_tahun.items():
    irisan = sorted(set(tahun_who) & set(tahun_bps))
    print(nama)
    print("Tahun irisan:", irisan)
    print("Jumlah tahun irisan:", len(irisan))
    print()
```

```python
# Mengubah metadata menjadi DataFrame
df_metadata = pd.DataFrame(metadata["value"])

# Kode UUID indikator WHO
kode_indikator = [
    "9A706FD",
    "A65146D",
    "217795A",
    "5C8435F"
]

# Menampilkan metadata khusus empat indikator
metadata_target = df_metadata[
    df_metadata["UrlName"].isin(kode_indikator)
]

print("Jumlah indikator ditemukan:", len(metadata_target))

display(metadata_target.T)
```

```python

# Membuat tabel profil seri pembanding WHO

profil_seri = {
    "UHC_INDEX_REPORTED": {
        "bps_pembanding": "var1402",
        "satuan": "Indeks (maksimum 100)",
        "nilai": "INDEX_N",
        "keputusan": "Kandidat pembanding konseptual; sama-sama berkaitan dengan cakupan layanan kesehatan."
    },
    "FINPROTECTION_CATA_TOT_10_POP": {
        "bps_pembanding": "var2279",
        "satuan": "Persentase populasi",
        "nilai": "PERCENT_POP_N",
        "keputusan": "Tidak dapat dibandingkan langsung; tidak memiliki tahun irisan dengan BPS."
    },
    "HWF_0001": {
        "bps_pembanding": "var1477",
        "satuan": "Jumlah dokter per 10.000 penduduk",
        "nilai": "RATE_PER_10000_N",
        "keputusan": "Kandidat pembanding kontekstual; cakupan dokter WHO belum tentu sama dengan kategori Dokter Umum BPS."
    },
    "HWF_0006": {
        "bps_pembanding": "var1477",
        "satuan": "Jumlah perawat dan bidan per 10.000 penduduk",
        "nilai": "RATE_PER_10000_N",
        "keputusan": "Kandidat pembanding kontekstual; satuan dan cakupan berbeda dari data jumlah tenaga kesehatan BPS."
    }
}

pasangan_bps = {
    "var1402": bps_1402,
    "var2279": bps_2279,
    "var1477": bps_1477
}

hasil_profil = []

for nama, info in profil_seri.items():
    df = data_who[nama]

    tahun_who = sorted(df["DIM_TIME"].dropna().unique())
    df_bps = pasangan_bps[info["bps_pembanding"]]
    tahun_bps = sorted(df_bps["tahun"].dropna().unique())
    irisan = sorted(set(tahun_who) & set(tahun_bps))

    # Mencari dimensi tambahan selain dimensi waktu
    kolom_dimensi = [
        kolom for kolom in df.columns
        if kolom.startswith("DIM_")
        and kolom not in [
            "DIM_TIME",
            "DIM_TIME_TYPE",
            "DIM_GEO_CODE_M49",
            "DIM_GEO_CODE_TYPE",
            "DIM_PUBLISH_STATE_CODE"
        ]
    ]

    rincian_dimensi = []
    for kolom in kolom_dimensi:
        kategori = df[kolom].dropna().unique().tolist()
        if len(kategori) > 0:
            rincian_dimensi.append(
                f"{kolom}: {len(kategori)} kategori"
            )

    hasil_profil.append({
        "seri_who": nama,
        "nama_indikator": df["IND_NAME"].iloc[0],
        "periode_mulai": min(tahun_who),
        "periode_akhir": max(tahun_who),
        "jumlah_observasi": len(df),
        "wilayah": ", ".join(df["GEO_NAME_SHORT"].dropna().unique()),
        "tingkat_data": ", ".join(df["DIM_GEO_CODE_TYPE"].dropna().unique()),
        "satuan": info["satuan"],
        "kolom_nilai": info["nilai"],
        "breakdown": "; ".join(rincian_dimensi) if rincian_dimensi else "Tidak ada breakdown",
        "bps_pembanding": info["bps_pembanding"],
        "tahun_irisan": ", ".join(map(str, irisan)) if irisan else "Tidak ada",
        "jumlah_tahun_irisan": len(irisan),
        "keputusan": info["keputusan"]
    })

tabel_profil = pd.DataFrame(hasil_profil)

display(tabel_profil)
```

```python

from pathlib import Path

folder_reports = Path("../reports")
folder_reports.mkdir(parents=True, exist_ok=True)

tabel_profil.to_csv(
    folder_reports / "profil_seri_pembanding.csv",
    index=False
)

print("Tabel profil berhasil disimpan.")
```

```python

# Menampilkan seluruh dimensi dan kategori setiap seri WHO

kolom_dimensi = [
    kolom for kolom in data_who["FINPROTECTION_CATA_TOT_10_POP"].columns
    if kolom.startswith("DIM_")
    and kolom not in [
        "DIM_TIME",
        "DIM_TIME_TYPE",
        "DIM_GEO_CODE_M49",
        "DIM_GEO_CODE_TYPE",
        "DIM_PUBLISH_STATE_CODE"
    ]
]

for kolom in kolom_dimensi:
    print(f"\n{kolom}")
    print(
        data_who["FINPROTECTION_CATA_TOT_10_POP"][kolom]
        .dropna()
        .value_counts()
        .to_string()
    )
```

```python

# Membuat tabel rincian breakdown Financial Protection

df_fin = data_who["FINPROTECTION_CATA_TOT_10_POP"]

rincian_breakdown = []

for kolom in kolom_dimensi:
    frekuensi = df_fin[kolom].value_counts(dropna=True)

    for kategori, jumlah in frekuensi.items():
        rincian_breakdown.append({
            "seri_who": "FINPROTECTION_CATA_TOT_10_POP",
            "dimensi": kolom,
            "kategori": kategori,
            "jumlah_observasi": jumlah
        })

tabel_breakdown = pd.DataFrame(rincian_breakdown)

display(tabel_breakdown)

tabel_breakdown.to_csv(
    folder_reports / "breakdown_finprotection.csv",
    index=False
)

print("Rincian breakdown berhasil disimpan.")
```
