```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd().parent

file_anomali = {
    "var1402": "calon_anomali_var1402.csv",
    "var2279": "calon_anomali_var2279.csv",
    "var1477": "calon_anomali_var1477.csv"
}

data_anomali = {}

for seri, nama_file in file_anomali.items():
    path = ROOT / "reports" / nama_file

    try:
        df = pd.read_csv(path)
    except pd.errors.EmptyDataError:
        df = pd.DataFrame()

    data_anomali[seri] = df

    print(f"\n{seri}")
    print("Jumlah calon anomali:", len(df))
    print("Kolom:", df.columns.tolist())

    if not df.empty:
        display(df.head())
```

```python
# Membaca data Silver BPS
silver = pd.read_csv(ROOT / "data" / "silver" / "silver.csv")

bps = silver[
    silver["seri_kode"].isin(["var1402", "var1477", "var2279"])
].copy()

# Mengubah periode dan nilai menjadi numerik
bps["periode"] = pd.to_numeric(bps["periode"], errors="coerce")
bps["nilai"] = pd.to_numeric(bps["nilai"], errors="coerce")

# Tidak memasukkan agregat Indonesia
bps = bps[bps["wilayah_kode"] != 9999].copy()

# Ringkasan var1402 per tahun
ringkasan_1402 = (
    bps[bps["seri_kode"] == "var1402"]
    .groupby("periode")["nilai"]
    .agg(["count", "median", "mean", "min", "max"])
    .round(2)
)

ringkasan_1402["perubahan_median"] = (
    ringkasan_1402["median"].pct_change() * 100
).round(2)

display(ringkasan_1402)
```

```python
# Melihat nama kolom pada data var1477
data_1477 = bps[bps["seri_kode"] == "var1477"]

print("Kolom:", data_1477.columns.tolist())
display(data_1477.head())
```

Analisis perubahan var1477

```python
ringkasan_1477 = (
    bps[bps["seri_kode"] == "var1477"]
    .groupby(["rincian", "periode"])["nilai"]
    .agg(["count", "median", "mean", "min", "max"])
    .round(2)
    .reset_index()
)

ringkasan_1477["perubahan_median"] = (
    ringkasan_1477
    .groupby("rincian")["median"]
    .pct_change() * 100
).round(2)

display(ringkasan_1477)
```

Periksa perubahan serentak var1402

```python
data_1402 = bps[bps["seri_kode"] == "var1402"].copy()

perubahan_1402 = data_1402.pivot(
    index="wilayah_kode",
    columns="periode",
    values="nilai"
)

for tahun in [2021, 2022, 2023]:
    tahun_sebelumnya = tahun - 1

    perubahan = (
        (perubahan_1402[tahun] - perubahan_1402[tahun_sebelumnya])
        / perubahan_1402[tahun_sebelumnya] * 100
    )

    print(f"\nPerubahan {tahun_sebelumnya} ke {tahun}")
    print("Jumlah provinsi:", perubahan.notna().sum())
    print("Naik:", (perubahan > 0).sum())
    print("Turun:", (perubahan < 0).sum())
    print("Median perubahan (%):", round(perubahan.median(), 2))
```

Memeriksa kandidat var1402

```python
anomali_1402 = data_anomali["var1402"].copy()

data_1402 = bps[bps["seri_kode"] == "var1402"].copy()

pivot_1402 = data_1402.pivot(
    index="nama_wilayah",
    columns="periode",
    values="nilai"
)

def ambil_nilai(wilayah, tahun):
    if wilayah not in pivot_1402.index:
        return None

    if tahun not in pivot_1402.columns:
        return None

    return pivot_1402.loc[wilayah, tahun]

anomali_1402["nilai_sebelumnya"] = anomali_1402.apply(
    lambda x: ambil_nilai(x["wilayah"], x["tahun"] - 1),
    axis=1
)

anomali_1402["nilai_sesudahnya"] = anomali_1402.apply(
    lambda x: ambil_nilai(x["wilayah"], x["tahun"] + 1),
    axis=1
)

display(
    anomali_1402[
        ["wilayah", "tahun", "nilai", "nilai_sebelumnya",
         "nilai_sesudahnya", "batas_bawah", "batas_atas"]
    ]
)
```

```python
anomali_1477 = data_anomali["var1477"].copy()

data_1477 = bps[bps["seri_kode"] == "var1477"].copy()

pivot_1477 = data_1477.pivot(
    index=["nama_wilayah", "rincian"],
    columns="periode",
    values="nilai"
)

def ambil_nilai_1477(provinsi, kategori, tahun):
    key = (provinsi, kategori)

    if key not in pivot_1477.index or tahun not in pivot_1477.columns:
        return None

    return pivot_1477.loc[key, tahun]

anomali_1477["nilai_sebelumnya"] = anomali_1477.apply(
    lambda x: ambil_nilai_1477(
        x["provinsi"], x["kategori"], x["tahun"] - 1
    ),
    axis=1
)

anomali_1477["nilai_sesudahnya"] = anomali_1477.apply(
    lambda x: ambil_nilai_1477(
        x["provinsi"], x["kategori"], x["tahun"] + 1
    ),
    axis=1
)

display(anomali_1477[
    ["kategori", "provinsi", "tahun", "nilai",
     "nilai_sebelumnya", "nilai_sesudahnya",
     "batas_bawah", "batas_atas"]
])
```

```python
# Menghitung perubahan nilai antarperiode var1402
data_1402 = bps[bps["seri_kode"] == "var1402"].copy()

data_1402 = data_1402.sort_values(
    ["nama_wilayah", "periode"]
)

data_1402["nilai_sebelumnya"] = (
    data_1402.groupby("nama_wilayah")["nilai"].shift(1)
)

data_1402["perubahan"] = (
    data_1402["nilai"] - data_1402["nilai_sebelumnya"]
)

data_1402["persen_perubahan"] = (
    data_1402["perubahan"] /
    data_1402["nilai_sebelumnya"] * 100
)

ringkasan = (
    data_1402.groupby("periode")
    .agg(
        jumlah_provinsi=("nama_wilayah", "nunique"),
        median_perubahan=("persen_perubahan", "median"),
        provinsi_naik=("perubahan", lambda x: (x > 0).sum()),
        provinsi_turun=("perubahan", lambda x: (x < 0).sum())
    )
)

display(ringkasan)
```
